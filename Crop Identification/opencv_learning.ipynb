```python
import cv2

image = cv2.imread("sample.jpg")

if image is not None:
    grayscale_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    action = input("What you want to do with image(show or save)?")
    if action == "show":
        cv2.imshow("grayscale image", grayscale_image)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    if action == "save":
        name = input("enter saving name of image: ")
        cv2.imwrite(name+".jpg", grayscale_image)
        print(f"image saved with name {name}.jpg....")
else:
    print("Error: Image not loaded.")
```

```python
import cv2

path = input("Enter path of image which you want to open: ")
image = cv2.imread(path)

if image is not None:
    action = input("Enter what you want to draw(line, circle, rectangle, text): ")
    match action:
        case "line":
            pt1Str = input("Enter string point of line (x and y saperated with coma): ").split(',')
            pt1 = tuple(map(int, pt1Str))
            pt2Str = input("Enter ending point of line (x and y saperated with coma): ").split(',')
            pt2 = tuple(map(int, pt2Str))
            new_image = cv2.line(image, pt1, pt2, (0,0,0), 3)
            cv2.imshow("Image with line", new_image)
            saveAction = input("Do you want to save this updated image? ")
            if saveAction == "Yes":
                cv2.imwrite("image_with_line.jpg", new_image)
                print("Image saved with name image_with_line.jpg....")
            cv2.waitKey(0)
            cv2.destroyAllWindows()
        case "rectangle":
            pt1Str = input("Enter string corner point of rectangle (x and y saperated with coma): ").split(',')
            pt1 = tuple(map(int, pt1Str))
            pt2Str = input("Enter ending corner point of rectangle (x and y saperated with coma): ").split(',')
            pt2 = tuple(map(int, pt2Str))
            new_image = cv2.rectangle(image, pt1, pt2, (0,0,0), 3)
            cv2.imshow("Image with rectangle", new_image)
            saveAction = input("Do you want to save this updated image? ")
            if saveAction == "Yes":
                cv2.imwrite("image_with_rectangle.jpg", new_image)
                print("Image saved with name image_with_rectangle.jpg....")
            cv2.waitKey(0)
            cv2.destroyAllWindows()
        case "circle":
            pt1Str = input("Enter center point of circle (x and y saperated with coma): ").split(',')
            pt1 = tuple(map(int, pt1Str))
            radius = int(input("Enter radius of circle: "))
            new_image = cv2.circle(image, pt1, radius, (0,0,0), 3)
            cv2.imshow("Image with circle", new_image)
            saveAction = input("Do you want to save this updated image? ")
            if saveAction == "Yes":
                cv2.imwrite("image_with_circle.jpg", new_image)
                print("Image saved with name image_with_circle.jpg....")
            cv2.waitKey(0)
            cv2.destroyAllWindows()
        case "text":
            pt1Str = input("Enter cordinate point of text (x and y saperated with coma): ").split(',')
            pt1 = tuple(map(int, pt1Str))
            new_image = cv2.putText(image, "Hello, cotton!", pt1, cv2.FONT_HERSHEY_SCRIPT_SIMPLEX, 1.2, (0,0,0), 3)
            cv2.imshow("Image with text", new_image)
            saveAction = input("Do you want to save this updated image? ")
            if saveAction == "Yes":
                cv2.imwrite("image_with_text.jpg", new_image)
                print("Image saved with name image_with_text.jpg....")
            cv2.waitKey(0)
            cv2.destroyAllWindows()
        case _:
            print("Enter valid action...")
else: 
    print("Error: Image not loaded.")
```
